```python
# simulate 
# 简单的环境表示
env = {
    '+': lambda a, b: a + b,
    '-': lambda a, b: a - b,
    '*': lambda a, b: a * b,
    '/': lambda a, b: a / b,
}

# 自定义 eval 函数
def eval(exp, env):
    if isinstance(exp, (int, float)):
        return exp
    elif isinstance(exp, str):
        return env[exp]
    else:
        func = eval(exp[0], env)
        args = [eval(arg, env) for arg in exp[1:]]
        return func(*args)

# 自定义 apply 函数
def apply(func, *args):
    return func(*args)

# 测试
exp = ['+', 1, ['*', 2, 3]]
print(eval(exp, env))  # 输出: 7
```

```python
class Amb:
    def __init__(self):
        self.choices = []
        self.fail = False

    def choose(self, *choices):
        self.choices.append(choices)
        if self.fail:
            self.fail = False
            return self.backtrack()
        return choices[0]

    def backtrack(self):
        while self.choices:
            choices = self.choices.pop()
            if len(choices) > 1:
                self.choices.append(choices[1:])
                return choices[1]
        raise Exception("No more choices")

    def fail(self):
        self.fail = True

    def run(self, func):
        try:
            return func()
        except Exception as e:
            return None

# 示例：求解简单的约束满足问题
def example():
    amb = Amb()

    def solve():
        x = amb.choose(1, 2, 3)
        y = amb.choose(1, 2, 3)
        if x + y != 4:
            amb.fail()
            # print('fail')
        return x, y

    result = amb.run(solve)
    print(result)

example()
```
